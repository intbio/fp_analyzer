# FP Analyzer — interactive notebook

Reproducible analysis of fluorescence-polarization (FP) saturation-binding
experiments: reads raw plate-reader output, applies instrument corrections
(G-factor, blank subtraction, Q-factor), fits direct-binding models, computes a
Z'-factor, and exports results.

**How to use.** Run the cells top to bottom. Each step shows an interactive
panel — set the options and click the button. All analysis code lives in the
companion module `fp_core.py`; the notebook stays clean. To inspect the code,
open `fp_core.py` or run the optional cell at the very bottom.

On **myBinder** the bundled example files (`layout_dCas9.csv`, `data_dCas9.csv`)
load automatically, so you can click straight through. Replace them with your
own files in Step 2 to analyse your data.

**References**: Lakowicz (2006); Roehrl et al. (2004) *Biochemistry* 43, 16056;
Jarmoskaite et al. (2020) *eLife* 9, e57264; Zhang et al. (1999) *J. Biomol. Screen.* 4, 67.

## Step 1 — Setup

Imports the analysis module. Keep `fp_core.py` in the same folder as this notebook.

```python
import fp_core
from fp_core import (show_load_panel, show_qc_panel, show_fit_panel,
                     show_summary_panel, show_export_panel, get_analyzer)
print('Ready.')
```

## Step 2 — Load data

Attach a **layout** file (`well, condition, sample_type, concentration,
replicate`) and a **data** file (custom `well, parallel, perpendicular` CSV or a
native BMG Labtech FP export — autodetected). If you attach nothing, the bundled
example files are used.

Parameters: **[L*] nM** = total labelled-ligand concentration (default 0.5);
**A_ref** = free-fluorophore anisotropy (default 0.035); **G override** = tick
only to set G by hand. Technical read-cycles are averaged per well
automatically. Click **Load & process**.

```python
show_load_panel()
```

## Step 3 — Quality control

Click **Show QC** for the G-factor calibration, the per-condition report, the
Z'-factor table, and the raw titration curves. **Log X axis** toggles the
concentration scale.

```python
show_qc_panel()
```

## Step 4 — Fit a binding model

Models: `fp_quadratic` (recommended — exact, accounts for ligand depletion),
`one_site` (simple hyperbola), `hill` (cooperativity coefficient n). Every
parameter can be varied or fixed, with optional bounds; by default `Kd` and
`A_bound` vary while `A_free`, `Q`, and `Dl` are fixed from the QC values.

The fit runs on the individual replicate wells (technical cycles averaged), so
the reported standard errors reflect the true number of independent
measurements. The full native **lmfit report** (including R²) is shown after
each fit. **Log X axis** and **Y = fraction bound** toggle the plot axes.

```python
show_fit_panel()
```

## Step 5 — Summary

Click **Refresh** to compare K_D across conditions and view the overlaid
saturation curves. Pick a condition in **Scatchard** to see its post-fit
Scatchard plot (B/F vs B, from the fitted A_bound/A_free) — a diagnostic of
deviation from 1:1 binding, not an independent K_D. **Y = fraction bound** and
**Log X axis** toggle the axes.

```python
show_summary_panel()
```

## Step 6 — Export

Select tables and figures and click **Export ZIP** for a download link. Saturation plots are exported in both anisotropy and fraction-bound forms; Scatchard plots are exported per fitted condition.

```python
show_export_panel()
```
