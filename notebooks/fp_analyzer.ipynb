# FP Analyzer — interactive notebook

A reproducible pipeline for fluorescence-polarization (FP) saturation-binding
experiments: it reads raw plate-reader output, applies instrument corrections
(G-factor, blank subtraction, Q-factor), fits direct-binding models, computes a
Z'-factor for assay quality, and exports results.

**How to use**

Run the cells top to bottom. Each step shows an interactive panel — fill it in
and click the button. All analysis code lives in the companion module
`fp_core.py`, so the notebook itself stays clean. To inspect or edit the code,
open `fp_core.py` (see the optional cell at the very bottom).

**Steps**

1. Setup — import the module.
2. Load data — attach a layout file and a data file, set assay parameters.
3. Quality control — review G, Q, A_free, blanks, Z'-factor.
4. Fit — run a binding model on each condition.
5. Summary — compare K_D values; view the post-fit Scatchard plot.
6. Export — download a ZIP with all tables and figures.

**References**: Lakowicz (2006); Roehrl et al. (2004) *Biochemistry* 43, 16056;
Jarmoskaite et al. (2020) *eLife* 9, e57264; Zhang et al. (1999) *J. Biomol. Screen.* 4, 67.

## Step 1 — Setup

Run this once. It imports everything from `fp_core.py`. Keep `fp_core.py` in the
same folder as this notebook.

```python
import fp_core
from fp_core import (show_load_panel, show_qc_panel, show_fit_panel,
                     show_summary_panel, show_export_panel, get_analyzer)
print('Ready.')
```

## Step 2 — Load data

Attach the **layout** file (`well, condition, sample_type, concentration,
replicate`) and the **data** file (custom `well, parallel, perpendicular` CSV
or a native BMG Labtech FP export — autodetected). Set:

* **[L*] nM** — total labelled-ligand concentration (fixed).
* **A_ref** — free-fluorophore anisotropy (default 0.035) for the G-factor.
* **G override** — tick only to supply G manually.
* **Cycle** — which read cycle to use for BMG exports (default: last).

Click **Load & process**.

```python
show_load_panel()
```

## Step 3 — Quality control

Click **Show QC**. Review:

* `G ± SE` — should be ~0.7–1.3 with small SE.
* `Q` — near 1 means the probe quantum yield is unchanged on binding.
* `A_free ± SE` — from substrate wells; should match the lowest titration points.
* `Z'-factor` — > 0.5 excellent, 0–0.5 marginal, ≤ 0 unreliable.

The **Log X axis** checkbox switches the titration-overview plot between a
semilog and a linear concentration axis.

```python
show_qc_panel()
```

## Step 4 — Fit a binding model

* `fp_quadratic` (recommended) — exact solution accounting for ligand depletion.
* `one_site` — simple hyperbola; valid when [L*] ≪ K_D.
* `hill` — phenomenological, with a Hill coefficient n.

`A_free`, `Q`, and `Dl` are fixed automatically from the QC values for every
model; `Kd` and `A_bound` are free by default. Use **Run fit** for one
condition or **Fit all conditions** for all of them. The **Log X axis**
checkbox toggles the saturation-curve scale.

```python
show_fit_panel()
```

## Step 5 — Summary

Click **Refresh** to compare K_D across fitted conditions and to see the
overlaid saturation curves. Pick a condition in the **Scatchard** dropdown to
view its **post-fit Scatchard plot** (B/F vs B, built from the fitted A_bound
and A_free). This is a *diagnostic* of deviation from 1:1 binding — a straight
line is consistent with simple binding, curvature suggests multiple sites or
non-specific binding. It is not an independent K_D estimate. **Log X axis**
toggles the saturation-curve scale.

```python
show_summary_panel()
```

## Step 6 — Export

Select the tables and figures you want and click **Export ZIP** to get a
download link. Scatchard data and plots are exported per fitted condition.

```python
show_export_panel()
```
